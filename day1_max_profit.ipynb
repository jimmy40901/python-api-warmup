```python
def max_profit(prices: list[int]) -> int:
    #考慮串列為空的例外情況
    if len(prices) == 0:
        return 0
    
    min_price = prices[0] 
    max_profit = 0
    
    for price in prices:
        # 你的邏輯放這裡
        if price < min_price:
            min_price = price
        if max_profit < price - min_price:
            max_profit = price - min_price
    return max_profit


# 測試
assert max_profit([7, 1, 5, 3, 6, 4]) == 5  # 應該印出 5
assert max_profit([7, 6, 4, 3, 1]) == 0     # 應該印出 0
assert max_profit([]) == 0          # 補上這行：空陣列的邊界情況
assert max_profit([5]) == 0         # 也可以補這行：只有一天的邊界情況
print("Max Profit 全部測試通過！")
```
