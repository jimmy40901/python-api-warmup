```python
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}  # 字典：key=數字, value=索引
    
    for i in range(len(nums)):
        num = nums[i]
        s = target - num
        if s in seen:
            return [seen[s], i]
        else:
            seen[num] = i #將num存入字典，key=num, value=i
        # 你的邏輯放這裡
    
    return []  # 理論上不會走到這裡


# 測試
assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
assert two_sum([3, 3], 6) == [0, 1]
print("Two Sum 全部測試通過！")
```
